```python
# функции для сбора данных
ku=ccxt.kucoin()
by=ccxt.bybit()
def parsku(start_,end_):
    cans=[]
    since=start_
    while since<=end_:
        can=ku.fetch_ohlcv("BTC/USDT","15m",since,limit=1000)
        if not can:
            break
        c1=[c for c in can if c[0]<end_]
        if not c1:
            break
        cans.extend(c1)
        since = cans[-1][0] + 1  # Передвигаем `since` на последнюю свечу
        time.sleep(0.5) 
    return cans

def parsby(start_,end_):
    cans=[]
    since=start_
    while since<=end_:
        can=ku.fetch_ohlcv("BTC/USDT","15m",since,limit=1000)
        if not can:
            break
        c1=[c for c in can if c[0]<end_]
        if not c1:
            break
        cans.extend(c1)
        since = cans[-1][0] + 1  # Передвигаем `since` на последнюю свечу
        time.sleep(float(0.5)) 
    return cans


def normalize_k_better(k, threshold=150, slowdown=0.1):
    if k<=threshold:
        return k*0.9
    else:
        diff=k-threshold
        growth=diff/(1+slowdown*diff/95)
        return (threshold+growth)*0.9
    
```

```python
#дабавление данных и их подсчет
def standard_obv(close, volume):
    diff = close.diff()  # Разница цен
    sign = np.sign(diff)  # +1 (рост), -1 (падение), 0 (равно), NaN для первого
    adjusted_volume = sign * volume  # Умножаем на знак
    obv = adjusted_volume.cumsum().fillna(0)  # Кумулятивная сумма, NaN -> 0
    return obv
"""timestamp,  0
    open,      1  
    high,      2
    low,       3  
    close,     4  
    volume     5  """
columns=['timestamp','open','high','low','close','volue']
m15_pd=pd.DataFrame(m15,columns=columns)
rsi=ta.momentum.rsi(m15_pd['close'],14)
adx=ta.trend.adx(m15_pd['high'],m15_pd['low'],m15_pd['close'],14)
bb=ta.volatility.BollingerBands(m15_pd['close'],20,2)
bolban=bb.bollinger_pband()
obv=standard_obv(m15_pd['close'],m15_pd['volue'])
atr=ta.volatility.average_true_range(m15_pd['high'],m15_pd['low'],m15_pd['close'])
#добавление данных
m15_pd['rsi']=rsi
m15_pd['adx']=adx
m15_pd['bb']=bolban
#m15_pd['obv']=obv
m15_pd['atr']=atr
```
